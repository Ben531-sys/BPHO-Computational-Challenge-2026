```python
import random
import matplotlib.pyplot as plt
import math
from ipywidgets import interact_manual
import ipywidgets as widgets
```

```python
#Challenge 1: Random walk 2d

#MaxRotation is in both directions, eg. 0 is straight forward, while pi/4 is pi/4 to the left, and pi/4 to thr right max

def RandomWalk(Steps,StepDistance,MaxRotation):
    Cx=0
    Cy=0
    CurrentAngle=0
    global x
    global y
    x=[0]
    y=[0]
    for a in range(Steps*10**6):
        randomChoice=random.random()-0.5
        RandomRadian=CurrentAngle-randomChoice*MaxRotation
        CurrentAngle=RandomRadian
        dx=StepDistance*math.cos(RandomRadian)
        dy=StepDistance*math.sin(RandomRadian)
        Cx+=dx
        Cy+=dy
        x.append(Cx)
        y.append(Cy)

interact_manual(RandomWalk,
                Steps=widgets.IntSlider(value=1, min=0,max=100,step=1,description='Steps (10**6)'),
                StepDistance=widgets.IntSlider(value=1, min=1,max=1000,description='Step Length'),
                MaxRotation=widgets.FloatSlider(value=2*math.pi, min=0,max=2*math.pi,step=math.pi/4,description='Maximum Rotation'))
```

```python
fig, ax = plt.subplots(1, 1)
x_data = x
y_data = y

ax.grid(True,'both',"both")
ax.plot(x_data, y_data, linewidth=1)
plt.tight_layout()
ax.set_ylabel("Y position (m)")
ax.set_xlabel("X postion (m)")
ax.set_title("Random Walk")

minArray=[0,0]
maxArray=[0,0]
minArray[0] = min(x)
minArray[1] = min(y)
maxArray[0] = max(x)
maxArray[1] = max(y)
Smallest=min(minArray)
Largest=max(maxArray)

ax.set_xlim(Smallest, Largest)
ax.set_ylim(Smallest, Largest)

plt.show()
```
