## Task #1. Random walk

Create a model of a random walk of *N* steps of size *s*. Each step is in a random direction, with angle θ from the horizontal chosen from a uniform distribution between 0 and 2π radians.

```python
import math
import random

import ipywidgets as widgets
import matplotlib.pyplot as plt
from ipywidgets import interact
from matplotlib.collections import LineCollection
```

```python
# Challenge 1: Random walk 2d

random.seed(
    0
)  # fixed on first run so the notebook is reproducible from the commandline


def RandomWalk(Steps, StepDistance):
    Cx = 0
    Cy = 0
    x = [0]
    y = [0]
    for a in range(Steps):
        RandomRadian = random.uniform(0, 2 * math.pi)
        dx = StepDistance * math.cos(RandomRadian)
        dy = StepDistance * math.sin(RandomRadian)
        Cx += dx
        Cy += dy
        x.append(Cx)
        y.append(Cy)

    fig, ax = plt.subplots(1, 1)
    x_data = x
    y_data = y

    ax.grid(True, "both", "both")
    segments = [
        ((x_data[i], y_data[i]), (x_data[i + 1], y_data[i + 1]))
        for i in range(len(x_data) - 1)
    ]
    path = LineCollection(segments, cmap="viridis", linewidth=1)
    path.set_array(range(len(segments)))
    ax.add_collection(path)
    plt.tight_layout()
    ax.set_ylabel("Y position (m)")
    ax.set_xlabel("X postion (m)")
    ax.set_title("Random Walk")

    minArray = [0, 0]
    maxArray = [0, 0]
    minArray[0] = min(x)
    minArray[1] = min(y)
    maxArray[0] = max(x)
    maxArray[1] = max(y)
    Smallest = min(minArray)
    Largest = max(maxArray)

    ax.set_xlim(Smallest, Largest)
    ax.set_ylim(Smallest, Largest)

    plt.show()


interact(
    RandomWalk,
    Steps=widgets.IntSlider(
        value=2000, min=10, max=20000, step=10, description="N (steps)"
    ),
    StepDistance=widgets.IntSlider(
        value=1, min=1, max=1000, description="s (step size)"
    ),
)
```
