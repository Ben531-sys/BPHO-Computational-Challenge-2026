```python
from math import sin
from math import cos
from math import pi
from ipywidgets import FloatSlider
from ipywidgets import interact_manual
```

```python
def DegreesToRadians(Degrees):
    return pi*(Degrees/180)

def ClassicalCalc(Theta,OtherTheta):
    cosValue=(cos(Theta)**2)*(cos(OtherTheta)**2)
    sinValue=(sin(Theta)**2)*(sin(OtherTheta)**2)
    return 1-cosValue-sinValue

def QuantemCalc(Theta,OtherTheta):
    return sin(OtherTheta-Theta)**2

def expirement(Theta,OtherTheta):
    ThetaRad=DegreesToRadians(Theta)
    OtherThetaRad=DegreesToRadians(OtherTheta)
    print(f"Classical Mismatch: {round(ClassicalCalc(ThetaRad,OtherThetaRad),8)}, Quantem Mismatch: {round(QuantemCalc(ThetaRad,OtherThetaRad),2)}")

interact_manual(
    expirement,
    Theta=FloatSlider(value=-30,min=-360,max=360,step=0.1,description="Theta: (Degrees)"),
    OtherTheta=FloatSlider(value=30,min=-360,max=360,step=0.1,description="Phi: (Degrees)"),
)
```
