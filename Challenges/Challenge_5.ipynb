## Task #5. Hydrogen emission spectrum and Bohr's model of a hydrogenic atom

Create a graph of photon energy vs wavelength for photon emissions from hydrogen atoms due to transitions between electron energy levels.

```python
import matplotlib.pyplot as plt
```

```python
# photon emission Calc


def PCalcEnergy(a, max):
    ray = []
    h = 6.626 * 10**-34  # plancks constant
    c = 299792458  # speed of light
    e = 1.602176634 * 10**-19  # Elementry Charge of an Electon
    for x in range(max - a):
        ray.append(((h * c) / Calc(a, max - x)) / e)
    return ray


def PCalcLength(a, max):
    ray = []
    for x in range(max - a):
        ray.append(Calc(a, max - x) * 10**9)
    return ray


def Calc(m, n):
    ep = 8.854187817 * 10**-12  # epilson 0
    h = 6.626 * 10**-34  # plancks constant
    c = 299792458  # speed of light
    me = 9.109 * 10**-31  # mass of an electron
    e = 1.602176634 * 10**-19  # Elementry Charge of an Electon
    return (
        (8 * (ep**2) * (h**3) * c)
        / (me * (e**4))
        * (((1 / (m**2)) - (1 / (n**2))) ** -1)
    )


def Expirement(max):
    global LymanLength, LymanEnergy, BalmerLength, BalmerEnergy, PaschenLength, PaschenEnergy, BrackettLength, BrackettEnergy, PfundLength, PfundEnergy
    LymanLength = PCalcLength(1, max)
    BalmerLength = PCalcLength(2, max)
    PaschenLength = PCalcLength(3, max)
    BrackettLength = PCalcLength(4, max)
    PfundLength = PCalcLength(5, max)

    LymanEnergy = PCalcEnergy(1, max)
    BalmerEnergy = PCalcEnergy(2, max)
    PaschenEnergy = PCalcEnergy(3, max)
    BrackettEnergy = PCalcEnergy(4, max)
    PfundEnergy = PCalcEnergy(5, max)


Expirement(max=6)
```

```python
def Graph(Lyman, Balmer, Paschen, Brackett, Pfund):
    fig, ax = plt.subplots(1, 1)
    ax.grid(True, "both", "both")
    ax.set_ylabel("Photon Energy (eV)")
    ax.set_xlabel("Wavelength (nm)")
    ax.set_title("Hydrogen Atom Photon Emissions")

    if Lyman:
        LymanColor = "pink"
        ax.plot(
            LymanLength,
            LymanEnergy,
            "*",
            color=LymanColor,
            linestyle="",
            markersize=6,
            markeredgewidth=0.1,
            label="Lyman",
        )
        for length in LymanLength:
            ax.axvline(
                x=length, color=LymanColor, linestyle="--", alpha=0.4, linewidth=0.5
            )
    if Balmer:
        BalmerColor = "red"
        ax.plot(
            BalmerLength,
            BalmerEnergy,
            "*",
            color=BalmerColor,
            linestyle="",
            markersize=6,
            markeredgewidth=0.1,
            label="Balmer",
        )
        for length in BalmerLength:
            ax.axvline(
                x=length, color=BalmerColor, linestyle="--", alpha=0.4, linewidth=0.5
            )
    if Paschen:
        PaschenColor = "blue"
        ax.plot(
            PaschenLength,
            PaschenEnergy,
            "*",
            color=PaschenColor,
            linestyle="",
            markersize=6,
            markeredgewidth=0.1,
            label="Paschen",
        )
        for length in PaschenLength:
            ax.axvline(
                x=length, color=PaschenColor, linestyle="--", alpha=0.4, linewidth=0.5
            )
    if Brackett:
        BrackettColor = "green"
        ax.plot(
            BrackettLength,
            BrackettEnergy,
            "*",
            color=BrackettColor,
            linestyle="",
            markersize=6,
            markeredgewidth=0.1,
            label="Brackett",
        )
        for length in BrackettLength:
            ax.axvline(
                x=length, color=BrackettColor, linestyle="--", alpha=0.4, linewidth=0.5
            )
    if Pfund:
        PfundColor = "Black"
        ax.plot(
            PfundLength,
            PfundEnergy,
            "*",
            color=PfundColor,
            linestyle="",
            markersize=6,
            markeredgewidth=0.1,
            label="Pfund",
        )
        for length in PfundLength:
            ax.axvline(
                x=length, color=PfundColor, linestyle="--", alpha=0.4, linewidth=0.5
            )
    ax.set_xlim(left=0)
    ax.set_ylim(bottom=0)
    ax.legend()
    plt.show()


Graph(Lyman=True, Balmer=True, Paschen=True, Brackett=True, Pfund=True)
```
