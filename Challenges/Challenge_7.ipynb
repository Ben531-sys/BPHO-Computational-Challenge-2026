```python
import matplotlib.pyplot as plt
import ipywidgets
import math
import numpy
import ipywidgets as widgets
import time
import cmath
```

```python
#Energy
def graph(energy,quantemlevels,discreteEnergy,descreteQuatemLevels,a):
    fig, ax = plt.subplots(1, 1)
    ax.grid(True,'both',"both")
    ax.set_ylabel("Energy (eV)")
    ax.set_xlabel("Quantum Level (#)")
    ax.set_title(f"Energy VS Quantum Level, of a Particle in a box of width {a}m")
    ax.plot(quantemlevels,energy,linewidth=0.5)
    ax.scatter(descreteQuatemLevels,discreteEnergy,zorder=5,s=3)
    ax.set_xlim(0.5, max(quantemlevels))
    ax.set_ylim(0, max(energy) * 1.05)

def calculation(n,area):
    hbar=6.62607015*10**-34/(2*math.pi)
    m=9.1094*(10**-31)
    return (((hbar**2)*(math.pi**2)*(n**2))/(2*m*area**2))/(1.602*10**-19)

def expirment(area,maxquatemnumber):
    a=10**area
    quantemlevels1=numpy.linspace(0,maxquatemnumber,maxquatemnumber*10)

    quantemlevels2=numpy.linspace(1,maxquatemnumber,maxquatemnumber)

    energy1=calculation(quantemlevels1,a)
    energy2=calculation(quantemlevels2,a)

    graph(energy1,quantemlevels1,energy2,quantemlevels2,a)
    time.sleep(0.1)

ipywidgets.interact(
    expirment,
    area=widgets.FloatSlider(value=-1,min=-15,max=5,description="Length: 10**(x)"),
    maxquatemnumber=widgets.IntSlider(value=5,min=1,max=100,step=1),
)
```

```python
h=6.62607015*10**-34
m=9.1094*(10**-31)

def ECalc(n,a):
    hbar=h/(2*math.pi)
    return (((hbar**2)*(math.pi**2)*(n**2))/(2*m*a**2))

def FinalCalc(n,t,a,x):
    E=ECalc(n,a)
    return (numpy.abs(numpy.sqrt(2/a)*numpy.exp(-(1j*E*t/h))*numpy.sin(n*numpy.pi*x/a)))**2
    
def expirment(maxN,a,NumberOfXChecks):
    fig, ax = plt.subplots(1, 1)
    ax.grid(True,'both',"both")
    ax.set_ylabel("Probality Density (eV)")
    ax.set_xlabel("x (angstorms)")
    ax.set_title(f"Probality Density VS X position of a Particle In A Box of Length {round(a,3)} angstroms")

    ns=numpy.linspace(1,maxN,maxN)
    possibleXs=numpy.linspace(0,a,NumberOfXChecks)
    t=1
    maxValue=0
    for n in ns:
        ProbalityDensity=FinalCalc(n,t,a*(10**-10),possibleXs)
        maxValue=max([maxValue,max(ProbalityDensity)])
        Energy=ECalc(n,a*(10**-10))/(1.602176634*10**-19)
        ax.plot(possibleXs,ProbalityDensity,label=f"n={n}, E={Energy}")
    ax.legend(loc="upper right")
    ax.set_xlim(0,a)
    ax.set_ylim(0,maxValue*1.05)
    time.sleep(1)

ipywidgets.interact(
    expirment,
    maxN=widgets.IntSlider(value=3,min=1,max=10,step=1),
    a=widgets.FloatSlider(value=0.1,min=0.01,step=0.01,max=1,description="Angstroms"),
    NumberOfXChecks=widgets.IntSlider(value=100,min=100,max=1000,step=1)
)
```
