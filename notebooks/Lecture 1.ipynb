```python
from numpy import linspace, sqrt
import ipywidgets as widgets
from ipywidgets import interact
from ipywidgets import interact_manual

import matplotlib.pyplot as plt
import math
#Drop simulator
Gravity = 9.81 #m/s^2


MaxHeight=10 #m
MinHeight=0 #m
NumberOfDrops=1000 #num

def drop_simulator(MaxHeight, MinHeight=0):
    Heights = linspace(MinHeight, MaxHeight, NumberOfDrops)
    Times = sqrt(2 * Heights / Gravity)

    fig, ax = plt.subplots(1, 1)
    x_data = Heights
    y_data = Times
 
    ax.grid(True,'both',"both")
    ax.plot(x_data, y_data, linewidth=1)
    plt.tight_layout()
    ax.set_ylabel("Time (s)")
    ax.set_xlabel("Drop Height (m)")
    ax.set_title("Droping Objects: Time VS Drop Height (No Air resistance)")
    plt.xlim(left=0)
    plt.ylim(bottom=0)
    plt.show()


interact_manual(drop_simulator, MaxHeight=widgets.FloatSlider(value=10, min=0, max=20), MinHeight=widgets.FloatSlider(value=0, min=0, max=10))
```

```python
import matplotlib.pyplot as plt
import math
import random

#1 Random Walk
Steps=10**6
x,y=0,0
X=[0]
Y=[0]
StepLength=1

for a in range(Steps):
    RandomAroundCircle=random.random() #0 is 0^o 1 is 360^o
    RandomRadian=2*math.pi*RandomAroundCircle
    dx=StepLength*math.cos(RandomRadian)
    dy=StepLength*math.sin(RandomRadian)
    X.append(X[a]+dx)
    Y.append(Y[a]+dy)

fig, ax = plt.subplots(1, 1)
x_data = X
y_data = Y

ax.grid(True,'both',"both")
ax.plot(x_data, y_data, linewidth=1)
plt.tight_layout()
ax.set_ylabel("Y position (m)")
ax.set_xlabel("X postion (m)")
ax.set_title("Random Walk")

minArray=[0,0]
maxArray=[0,0]
minArray[0] = min(X)
minArray[1] = min(Y)
maxArray[0] = max(X)
maxArray[1] = max(Y)
Smallest=min(minArray)
Largest=max(maxArray)

ax.set_xlim(Smallest, Largest)
ax.set_ylim(Smallest, Largest)

plt.show()
```

```python
import matplotlib.pyplot as plt
import math
import random

#Num Random Walk

def Walk(Steps,StepLength):
    x,y=0,0
    X=[0]
    Y=[0]

    for a in range(Steps):
        RandomAroundCircle=random.random() #0 is 0^o 1 is 360^o
        RandomRadian=2*math.pi*RandomAroundCircle
        dx=StepLength*math.cos(RandomRadian)
        dy=StepLength*math.sin(RandomRadian)
        X.append(X[a]+dx)
        Y.append(Y[a]+dy)
    return(X,Y)

StepLength=1
NumberOfWalks=50
Steps=10**6
fig, ax = plt.subplots(1, 1)
ax.grid(True,'both',"both")
minArray=[]
maxArray=[]

for p in range(NumberOfWalks):
    X,Y=Walk(Steps,StepLength)
    Colour=[random.random(),random.random(),random.random()]
    ax.plot(X, Y, linewidth=1,c=Colour)
    minArray.append(min(X))
    minArray.append(min(Y))
    maxArray.append(max(X))
    maxArray.append(max(Y))

plt.tight_layout()
ax.set_ylabel("Y position (m)")
ax.set_xlabel("X postion (m)")
ax.set_title("Lots of Random Walk")

Smallest=min(minArray)
Largest=max(maxArray)

ax.set_xlim(Smallest, Largest)
ax.set_ylim(Smallest, Largest)

plt.show()
```
