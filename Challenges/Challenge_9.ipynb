## Task #9. Compton scattering

Plot fractional wavelength shift Δλ/λ, electron recoil speed *v* and electron recoil angle φ vs photon scattering angle θ.

```python
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, interact
from numpy import arctan, cos, degrees, linspace, pi, sin, sqrt
```

```python
c = 2.99792458 * 10**8  # speed of light (m/s)
h = 6.62607015 * 10**-34  # plancks constant (J s)
ME = 9.1093837 * 10**-31  # electron mass (kg)
e = 1.602176634 * 10**-19  # electron charge (C)


def DeltaLamda(ScatteringAngle):
    return (h / (ME * c)) * (1 - cos(ScatteringAngle))


def FractionalWaveShift(ScatteringAngle, EnergyJoules):
    deltaLamda = DeltaLamda(ScatteringAngle)
    Lamda = (h * c) / EnergyJoules
    return deltaLamda / Lamda


def Whiplash(ScatteringAngle, EnergyJoules):
    Lamda = (h * c) / EnergyJoules
    NewLamda = Lamda + DeltaLamda(ScatteringAngle)
    ValueX = ME * c**2
    return sqrt(1 - (ValueX / (h * c * (1 / Lamda - 1 / NewLamda) + ValueX)) ** 2)


def ElectronRecoil(Angle, EnergyJoules):
    Lamda = (h * c) / EnergyJoules
    alpha = h / (ME * c * Lamda)
    denominator = (1 + alpha) * (1 - cos(Angle))
    ReturnAngle = arctan(sin(Angle) / denominator)
    return degrees(ReturnAngle)


def expirement(MaxEnergyLog, EnergyLevelsCount, MaxAngle):
    fig, ax = plt.subplots(1, 3, figsize=(14, 5))

    MaxEnergyKeV = 10**MaxEnergyLog

    EnergyLevelsKeV = linspace(10, MaxEnergyKeV, EnergyLevelsCount, dtype=int)
    Angles = linspace(0, MaxAngle * (pi / 180), 500)

    for energy_kev in EnergyLevelsKeV:
        EnergyJoules = energy_kev * 1000 * e

        ax[0].plot(
            Angles,
            FractionalWaveShift(Angles, EnergyJoules),
            label=f"E={energy_kev}keV",
        )
        ax[1].plot(Angles, Whiplash(Angles, EnergyJoules), label=f"E={energy_kev}keV")
        ax[2].plot(
            Angles, ElectronRecoil(Angles, EnergyJoules), label=f"E={energy_kev}keV"
        )

    for a in ax:
        a.grid(True, which="both", linestyle="--", alpha=0.6)
        a.set_xlabel("Photon Scattering Angle (radians)")
        a.set_xlim(0)

    ax[0].set_ylabel("λ / λ")
    ax[1].set_ylabel("Electron Recoil Speed (v/c)")
    ax[2].set_ylabel("Electron Recoil Angle (ɸ/deg)")

    ax[0].set_title("Fractional Wave Shift")
    ax[1].set_title("Recoil Speed ratio")
    ax[2].set_title("Recoil Angle")

    fig.suptitle("Compton Effect Simulation", fontsize=14, y=1.02)
    ax[2].legend(loc="upper right")

    ax[1].set_ylim(0, 1)
    ax[0].set_ylim(0)

    plt.tight_layout()
    plt.show()


interact(
    expirement,
    MaxEnergyLog=IntSlider(value=3, min=1, max=10),
    EnergyLevelsCount=IntSlider(value=3, min=1, max=10),
    MaxAngle=IntSlider(value=150, min=1, max=359, description="Max Angle (Degrees): "),
)
```
