## Task #8. Quantum cryptography

Create a visual calculator of the classical and quantum mismatch probabilities for the detection of polarized entangled photons.

```python
from math import cos, pi, sin

import matplotlib.pyplot as plt
import numpy
from ipywidgets import FloatSlider, interact
```

```python
def DegreesToRadians(Degrees):
    return pi * (Degrees / 180)


def ClassicalCalc(Theta, OtherTheta):
    cosValue = (cos(Theta) ** 2) * (cos(OtherTheta) ** 2)
    sinValue = (sin(Theta) ** 2) * (sin(OtherTheta) ** 2)
    return 1 - cosValue - sinValue


def QuantemCalc(Theta, OtherTheta):
    return sin(OtherTheta - Theta) ** 2


def expirement(Theta, OtherTheta):
    ThetaRad = DegreesToRadians(Theta)
    OtherThetaRad = DegreesToRadians(OtherTheta)

    PhiSweep = numpy.linspace(-360, 360, 721)
    PhiSweepRad = numpy.radians(PhiSweep)
    Classical = [ClassicalCalc(ThetaRad, p) for p in PhiSweepRad]
    Quantum = [QuantemCalc(ThetaRad, p) for p in PhiSweepRad]

    fig, ax = plt.subplots(figsize=(9, 5))
    ax.grid(True, alpha=0.5)
    ax.plot(PhiSweep, Classical, label="Classical (hidden-variable)", color="tab:blue")
    ax.plot(PhiSweep, Quantum, label="Quantum (entangled)", color="tab:orange")

    CurrentClassical = ClassicalCalc(ThetaRad, OtherThetaRad)
    CurrentQuantum = QuantemCalc(ThetaRad, OtherThetaRad)
    ax.plot(OtherTheta, CurrentClassical, "o", color="tab:blue", markersize=8)
    ax.plot(OtherTheta, CurrentQuantum, "o", color="tab:orange", markersize=8)

    ax.set_xlabel("Phi: OtherTheta (Degrees)", fontsize=11)
    ax.set_ylabel("Mismatch Probability", fontsize=11)
    ax.set_title(
        f"Mismatch Probability vs Phi (Theta = {Theta}°)",
        fontsize=12,
        fontweight="bold",
    )
    ax.set_ylim(-0.05, 1.05)
    ax.legend(loc="upper right", frameon=True)

    plt.show()

    print(
        f"Classical Mismatch: {round(CurrentClassical,8)}, Quantem Mismatch: {round(CurrentQuantum,2)}"
    )


interact(
    expirement,
    Theta=FloatSlider(
        value=-30, min=-360, max=360, step=0.1, description="Theta: (Degrees)"
    ),
    OtherTheta=FloatSlider(
        value=30, min=-360, max=360, step=0.1, description="Phi: (Degrees)"
    ),
)
```
